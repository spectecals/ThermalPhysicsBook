9. Path Independence (TP 11.3)

   Consider the differential:

   $$
   dz = 2xy \, dx + (x^2 + 2y) \, dy
   $$

   Evaluate the integral $\int^{(x_2, y_2)}_{(x_1, y_1)} dz$ along the paths consisting of the straight line segments:

   1. $(x_1, y_1) \rightarrow (x_2, y_1)$ and then $(x_2, y_1) \rightarrow (x_2, y_2)$.
   2. $(x_1, y_1) \rightarrow (x_1, y_2)$ and then $(x_1, y_2) \rightarrow (x_2, y_2)$.
   3. Is $dz$ an exact differential (i.e. does the integral depend only on its start and end points)?
