### Answers

1. Cloud Formation

   As air rises over a mountain it moves into a region of lower atmospheric pressure. The parcel expands and does work on its surroundings. During the ascent it is not in meaningful thermal contact with its surroundings, so we take the process to be adiabatic ($Q=0$).

   From the first law,

   $$
   \Delta U = Q - W.
   $$

   Since the gas expands, the work done *by* the gas is positive ($W>0$), so $\Delta U<0$ and the air cools.

   As the temperature drops, the saturation vapour pressure of water decreases. When the saturation vapour pressure falls below the water vapour partial pressure in the parcel, the vapour becomes supersaturated and condenses into droplets. Those droplets form clouds, which is why cloud formation often occurs near mountain tops.

2. Gas Compression

   1. Isothermal Compression ($T=\text{constant}$)

      For an ideal gas, internal energy depends only on temperature, so if $T$ is constant then $\Delta U=0$. The reversible work done *by* the gas is

      $$
      W = \int_{V_1}^{V_2} p\,dV.
      $$

      Using $p=\frac{nRT}{V}$ with $T$ constant,

      $$
      W = \int_{V_1}^{V_2}\frac{nRT}{V}\,dV
        = nRT\ln\!\left(\frac{V_2}{V_1}\right).
      $$

      For compression $V_2<V_1$, so $\ln(V_2/V_1)<0$ and hence $W<0$ (work done by the gas is negative; equivalently work done *on* the gas is positive).

   2. Adiabatic Compression ($Q=0$)

      For an adiabatic process,

      $$
      \Delta U = -W.
      $$

      During compression the work done *by* the gas is negative ($W<0$), so $\Delta U>0$ and the temperature rises.

      For a quasi-static adiabatic process, the work done by the gas can be written as

      $$
      W=\frac{p_2V_2 - p_1V_1}{\gamma-1}.
      $$

3. Increasing the Temperature of a Gas

   The ideal gas law is:

   $$
   pV=nRT,
   $$

   so raising $T$ requires a change in $p$, $V$, or both.

   - **Constant pressure (isobaric):** if $p=\text{constant}$ then $V\propto T$. On a $pV$ diagram this is a horizontal line moving to the right as $T$ increases from $T_c$ to $T_h$.

   - **Constant volume (isochoric):** if $V=\text{constant}$ then $p\propto T$. On a $pV$ diagram this is a vertical line moving upward as $T$ increases.

4. Isothermal Processes

   For a reversible isothermal process, the work done *by* the gas is

   $$
   W = nRT\ln\!\left(\frac{V_2}{V_1}\right).
   $$

   Here $n=1\ \text{mol}$, $T=300\ \text{K}$, and $V_2=2V_1$, so

   $$
   W = (1)(8.31)(300)\ln 2 \approx 1.73\times10^3\ \text{J}.
   $$

   This is positive (expansion). The work done *on* the system is therefore

   $$
   W_{\text{on}}=-W\approx -1.7\times10^3\ \text{J}.
   $$

5. Isothermal Expansion (TP 11.1)

   1. Why doesn’t the internal energy change?

      For an ideal gas the internal energy depends only on temperature (it is purely kinetic in the ideal-gas model). Since the process is isothermal at $T_0$, the temperature does not change and therefore $\Delta U=0$.

   2. Work done by the gas and heat flow into the gas

      Using $p=\frac{nRT_0}{V}$,

      $$
      W=\int_{V_1}^{V_2}p\,dV
       =\int_{V_1}^{V_2}\frac{nRT_0}{V}\,dV
       =nRT_0\ln\!\left(\frac{V_2}{V_1}\right).
      $$

      For $n=1$,

      $$
      W=RT_0\ln\!\left(\frac{V_2}{V_1}\right).
      $$

      Since $\Delta U=0$, the first law gives $0=Q-W$, so

      $$
      Q=W=RT_0\ln\!\left(\frac{V_2}{V_1}\right).
      $$

6. Heat Capacities and the Gas Constant (TP 11.2)

   Definitions:

   $$
   C_V=\left(\frac{1}{n}\frac{\partial Q}{\partial T}\right)_V,\qquad
   C_p=\left(\frac{1}{n}\frac{\partial Q}{\partial T}\right)_p,\qquad
   \gamma=\frac{C_p}{C_V}.
   $$

   For an ideal gas,

   $$
   C_p - C_V = R.
   $$

   Using $C_p=\gamma C_V$ in $C_p-C_V=R$ gives

   $$
   \gamma C_V - C_V = R
   \;\Rightarrow\;
   C_V(\gamma-1)=R
   \;\Rightarrow\;
   \frac{R}{C_V}=\gamma-1.
   $$

   Dividing $C_V(\gamma-1)=R$ by $C_p$:

   $$
   \frac{R}{C_p}=\frac{C_V(\gamma-1)}{C_p}
   =\frac{\gamma-1}{\gamma}.
   $$

7. Cyclic Processes (UPMP Example 19.3)

   1. Why is the work negative?

      The net work done *by* the system over a cycle is the enclosed area:

      $$
      W=\oint p\,dV.
      $$

      A counterclockwise cycle corresponds to compression occurring (on average) at higher pressure than expansion, so the surroundings do more work on the system than the system does on the surroundings. Hence the net work done *by* the system is negative.

   2. Change in internal energy and heat added

      Over a complete cycle the system returns to its initial state, so the change in internal energy is

      $$
      \Delta U=0.
      $$

      Using the first law $\Delta U=Q-W$ gives $0=Q-W$, so $Q=W$.

      With $W=-500\ \text{J}$ (work done by the system),

      $$
      Q=-500\ \text{J}.
      $$

      So 500 J of heat is removed from the system over the cycle.

8. Isobaric Boiling (UPMP Example 19.5)

   1. Work done *by* the water

      Convert volumes: $1\ \text{cm}^3=10^{-6}\ \text{m}^3$.

      $$
      V_1=1\times10^{-6}\ \text{m}^3,\qquad
      V_2=1671\times10^{-6}\ \text{m}^3.
      $$

      Hence,

      $$
      \Delta V = (1671-1)\times10^{-6} = 1.670\times10^{-3}\ \text{m}^3.
      $$

      At constant pressure,

      $$
      W=p\,\Delta V=(1.013\times10^5)(1.670\times10^{-3})
      \approx 1.69\times10^2\ \text{J}.
      $$

      So $W\approx 169\ \text{J}$.

   2. Increase in internal energy

      Heat added during vaporisation:

      $$
      Q=mL_v=(1.0\times10^{-3})(2.256\times10^6)=2.256\times10^3\ \text{J}.
      $$

      First law:

      $$
      \Delta U = Q - W \approx 2256 - 169 \approx 2.09\times10^3\ \text{J}.
      $$

      So $\Delta U \approx 2.09\times10^3\ \text{J}$.

9. Path Independence (TP 11.3)

   Given

   $$
   dz = 2xy\,dx + (x^2+2y)\,dy,
   $$

   identify $M(x,y)=2xy$ and $N(x,y)=x^2+2y$.

   1. Path (a): $(x_1,y_1)\to(x_2,y_1)\to(x_2,y_2)$

      Segment 1 ($y=y_1$, $dy=0$):

      $$
      \int_{x_1}^{x_2} 2xy_1\,dx
      = y_1(x_2^2-x_1^2).
      $$

      Segment 2 ($x=x_2$, $dx=0$):

      $$
      \int_{y_1}^{y_2} (x_2^2+2y)\,dy
      = x_2^2(y_2-y_1) + (y_2^2-y_1^2).
      $$

      Total:

      $$
      \int_{(a)} dz
      = x_2^2y_2 - x_1^2y_1 + (y_2^2-y_1^2).
      $$

   2. Path (b): $(x_1,y_1)\to(x_1,y_2)\to(x_2,y_2)$

      Segment 1 ($x=x_1$, $dx=0$):

      $$
      \int_{y_1}^{y_2} (x_1^2+2y)\,dy
      = x_1^2(y_2-y_1) + (y_2^2-y_1^2).
      $$

      Segment 2 ($y=y_2$, $dy=0$):

      $$
      \int_{x_1}^{x_2} 2xy_2\,dx
      = y_2(x_2^2-x_1^2).
      $$

      Total:

      $$
      \int_{(b)} dz
      = x_2^2y_2 - x_1^2y_1 + (y_2^2-y_1^2).
      $$

   Since the two paths give the same result, the integral is path independent.

   To test exactness, check

   $$
   \frac{\partial M}{\partial y}=\frac{\partial}{\partial y}(2xy)=2x,\qquad
   \frac{\partial N}{\partial x}=\frac{\partial}{\partial x}(x^2+2y)=2x.
   $$

   These are equal, so $dz$ is an exact differential and the integral depends only on the endpoints.
