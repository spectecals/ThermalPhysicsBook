      $$

   i. Isothermal expansion at $T_h$

   ii. Adiabatic expansion whilst temperature falls to $T_c$

   iii. Isothermal compression at $T_c$

   iv. Adiabatic compression while temperature rises to $T_h$

   1. Draw these stages on a labelled $PV$ diagram. (6)
   2. By considering the work output and the heat supplied, show that the efficiency of the Carnot cycle is given by (4)
      $$
      \eta = 1 - \left(\frac{Q_c}{Q_h}\right).
      $$
