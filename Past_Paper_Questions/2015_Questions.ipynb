3. Probability

   You are trying to pick the ace of hearts out of a pack of 52 shuffled cards. You pick a card at random to check if it is the card you want. If it is the ace of hearts, you are happy, and you stop. If it is not the ace of hearts, you return the card, reshuffle the pack, and try again. Your friend suggests that to successfully find the card you need to try taking out a card on average 52 times.

   1. Show that the probability of success based on her logic is $64\%$. Hint: you need to consider the probability of not succeeding at picking out the right card. (6)
