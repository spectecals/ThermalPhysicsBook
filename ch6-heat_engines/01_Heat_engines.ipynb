Considering the first law of thermodynamics applied at the case of thermodynamic cyclic processes, this means that: 

$$

U_{2}-U_{1}=0\Rightarrow Q=W,

$$

 i.e. the net heat flowing in heat engines equals the net work done.
All heat engines absorb heat $Q_{H}$ from a source at high temperature $T_{H}$, and after performing some mechanical work, discard some heat $Q_{C}$ at a lower temperature $T_{C}$.

The net heat $Q$ per cycle is: 

$$

Q=Q_{H} + Q_{C}

$$

In the particular case of the heat engine, as also shown from {numref}`Heat_engine_schematic`, $Q_{C}$ is negative since it's trasferred from the system to the cold reservoir, while $Q_{H}$ is positive. The net heat then can be also written as:

$$

Q=Q_{H}-|Q_{C}|

$$

The net work $W$ per cycle is equal to the net heat since we are dealing with a cyclic process: 

$$

W=Q_{H}+Q_{C}=Q_{H}-|Q_{C}|

$$

A way to quantify how well the engine is able to convert heat into work is by calculating the thermal efficiency $e$: 

$$

e=\frac{W}{Q_{H}} =1+\frac{Q_{C}}{Q_{H}}=1-\frac{|Q_{C}|}{Q_{H}}

$$
 
Since the heat rejected to the enviroment $Q_{C}$ in a heat engine is always negative, the efficiency of a heat engine is always less than unity. This fundamental limitation is expressed by the **Kelvin-Planck statement of the second law of thermodynamics**, which states that no heat engine can convert all the heat it receives into work. A portion of the energy must always be rejected to a lower-temperature environment. This limitation is not caused by friction or other practical imperfections. Even an ideal heat engine cannot achieve 100% efficiency. The maximum possible efficiency depends only on the temperatures of the hot and cold reservoirs, a result that will be discussed in the context of the Carnot cycle.




 
 

## Carnot cycle
According to second law of thermodynamics, a heat engine with a 100% efficiency cannot exist.
What is then the maximum conversion from heat to work reached by a heat engine?

The answer was given in 1824 by the French engineer Sadi Carnot, who 
developed a hypothetical, idealized heat engine that has the maximum possible efficiency  consistent with the second law. The cycle of this engine is called the Carnot cycle.

For maximum heat engine efficiency all processes must be reversible. This results in:
-  No temperature drops, i.e. isothermal processes at $T_{H}$ and $T_{C}$;
- When $T_{C}<T<T_{H}$, there is no heat transfer between heat engines and reservoirs, i.e. adiabatic processes occur.

```{figure} figures_for_chapter_6/CarnotFridge.png
:width: 60%
:name: Carnot_pv
$pV$ diagram of the Carnot engine cycle operating between $T_{H}$ and $T_{C}$. Blue curves correspond to isothermal processes, while red ones correspond to adiabatic transformations.
```

The Carnot cycle consists of four reversible processes: two isothermal processes and two adiabatic processes. The thermodynamic states of the Carnot cycle are illustrated in the $pV$ diagram shown in {numref}`Carnot_pv`. The cycle proceeds through the following four reversible processes:

1) **Process $a-b$: Reversible isothermal expansion**. The working gas is palced in contact with a hot reservoir at temperature $T_{H}$. The gas expands slowly while absorbing heat $Q_{H}$ from the reservoir. Because the temperature difference between the gas and the reservoir is infinitesimally small, the process is reversible and occurs at constant temperature.
2)  **Process $b-c$: Reversible adiabatic expansion**. The thermal contact with the reservoir is removed and the cylinder is thermally insulated. The gas continues to expand while performing work on the surroundings. Since no heat is exchanged during this process, the temperature of the gas decreases from $T_{H}$ to $T_{C}$.
3)  **Process $c-d$: Reversible isothermal compression**. The gas is brought into contact with the cold reservoir at temperature $T_{C}$. The surroundings compress the gas while heat $Q_{C}$ is rejected to the reservoir. The compression occurs at constant temperature.
4)  **Process $d-a$: Reversible adiabatic compression**. The gas is again thermally isolated and compressed further until its temperature increases from $T_{C}$ and $T_{H}$, returning the system to its initial state and completing the cycle.

On the $pV$ diagram the area enclosed by the cycle represents the net work delivered by the heat engine. The upper portion of the cycle corresponds to expansion processes where work is produced, while the lower portion corresponds to compression processes where work is done on the working fluid.

### The reversed Carnot cycle

Because all processes of the Carnot cycle are reversible, the entire cycle can be reversed. When the direction of the cycle is reversed, the device operates as a refriegerator or heat pump. In this case, heat $Q_{C}$ is extracted from the cold reservoir, heat $Q_{H}$ is rejected to the hot reservoir, and a net work input $W$ is required to drive the cycle. This reversed operation corresponds to the cycle shown in {numref}`Carnot_fridge_scheme` and {numref}`Carnot_fridge_pv`.

```{figure} figures_for_chapter_6/RefrigeratorSchematic.png
:width: 60%
:name: Carnot_fridge_scheme
General schematic of a Carnot fridge operating between $T_{H}$ and $T_{C}$.
```

Since the work produced by the super-Carnot engine matches the work required by the fridge to work, the resulting system will consist of a transfer of heat from the cold reservoir to the hot one without any work required. 

Moreover, from the definition of efficiency of the Carnot fridge given above, it is possible to observe:

$$

e_{CF}=\frac{Q_{C}}{W},\quad e_{CF}=\frac{Q_{C}^{CF}}{W(1+x)}\\
\Rightarrow Q_{C}^{CF}=Q_{C}(1+x)=\left(Q_{H}-W\right)(1+x)

$$

and then the heat transferred to the hot reservoir by the Carnot fridge will be:

$$

Q_{H}^{CF}=Q_{C}(1+x)+W(1+x)=\underbrace{(Q_{C}+W)}_{Q_{H}}(1+x)

$$

Let us express now also the total amount of heat rejected to the hot reservoir:

$$

Q_{H}-Q_{H}^{CF}=Q_{H}-Q_{H}(1+x)=-Q_{H}x,

$$

while the total heat extracted from the cold reservoir to the system will correspond

$$

-Q_{C}^{SC}+Q_{C}^{CF}=\underbrace{-Q_{H}+W(1+x)}_{-Q_{C}^{SC}}+\underbrace{\left(Q_{H}-W\right)\left(1+x\right)}_{Q_{C}^{CF}}=Q_{H}x

$$

Calculating the variation of entropy:

$$

\Delta S=\Delta S_{H}+\Delta S_{C}=+\frac{Q_{H}x}{T_{H}}-\frac{Q_{H}x}{T_{C}}=xQ_{H}\left(\frac{1}{T_{H}}-\frac{1}{T_{C}}\right)<0,

$$

which is negative since $T_{H}>T_{C}$. This clearly violates the second law of thermodynamics and then it is false to assume a super Carnot engine to exist with $x>0$. To satisfy second law, $x\leq 0$, which means the Carnot engine is the most efficient heat engine.

### Efficiency of Carnot cycle
To calculate the efficiency of the Carnot engine let's calculate:

$$

Q_{H}=W_{ab}=nRT_{H}\mathrm{ln}\frac{V_{b}}{V_{a}}>0,

$$

while:

$$

Q_{C}=W_{cd}=nRT_{C}\mathrm{ln}\frac{V_{d}}{V_{c}}<0

$$

In adiabatic processes $bc$ and $da$ it holds:

$$

\begin{gather*}
T_{H}V{b}^{\gamma-1}=T_{C}V_{c}^{\gamma-1}\\
T_{C}V_{d}^{\gamma-1}=T_{H}V_{a}^{\gamma-1}
\end{gather*},

$$

which then results:

$$

\frac{V_{b}}{V_{a}}=\frac{V_{c}}{V_{d}}

$$

The thermal efficiency of the cycle is then:

$$

e=1+\frac{Q_{C}}{Q_{H}}=1-\frac{T_{C}}{T_{H}}\underbrace{\frac{\mathrm{ln}\frac{V_{c}}{V_{d}}}{\mathrm{ln}\frac{V_{b}}{V_{a}}}}_{=1}=1-\frac{T_{C}}{T_{H}}

$$

Equivalently, the coefficient of performance ($COP$) of the Carnot refrigerator can be proven to be:

$$

COP=\frac{Q_{C}}{|W|}=\frac{T_{C}}{T_{H}-T_{C}}

$$
, where now $W$ is negative (input work required to drive the cycle).

### Carnot Principles

Two fundamental conclusions follow from the second law of thermodynamics. First, the efficiency of any real (irreversible) heat engine is always lower than that of a reversible engine operating between the same two temperature reservoirs. Second, all reversible heat engines operating between the same temperature have exactly the same efficiency, regardless of the working fluid or the specific details of the cycle. These results, known as the Carnot principles, establish the Carnot cycle as the benchmark for the maximum achievable efficiency of heat engines.
The Carnot cycle is not a practical engine cycle and cannot be implemented exactly in real machines. Nevertheless, it plays a fundamental role in thermodynamics beacuse it establishes the maximum possible efficiency that any heat engine can achieve between two given temperature reservoirs. No real engine operating between the same temperatures can be more efficient than a Carnot engine.

### Otto cycle
The Otto cycle is composed of 4 main processes: two adiabatic  and two isochoric processes summarized in  {numref}`InternalCombustion`, and outlined below:
- **Step 0**: Intake stroke. Here the piston moves down, allowing gasoline-air mixture to enter the cylinder via the intake valve.
- **Step 1 ($a-b$)**: Compression stroke. The intake valve closes and the piston compresses the misture (adiabatic process).
- **Step 2 ($b-c$)**: Ignition. The spark plug ignites the mixture.$Q_{H}$ is added to the system by the burnt fuel. This is an isochoric step.
- **Step 3 ($c-d$)**: Power stroke. The hot mixture expands with pistons being pushed down. This is an adiabatic expansion step, in which the volume is expanded from $V$ back to $rV$.
- **Step 4 ($d-a$)**: Exhaust stroke. The exhaust valve opens and combustion residuals are expelled. The cylinder is ready for the next cycle. This is an isochoric process, where heat $Q_{C}$ id rejected to the cold reservoir. 

This idealized model of the thermodynamic processes in a gasoline engine is called Otto cycle and its diagram is reported in {numref}`Otto_pv`



Let us now calculate the efficiency of the Otto cycle

Let us start from the ioschoric processes $b-c$ and $d-a$. An isochoric process is a process at constant volume. Since the volume is not changed, o work is done, $W=0$, as:

$$

dW=pdV

$$

From the first law of thermodynamics $\Delta U=Q-W=Q$ and then:

$$

Q=nC_{V}\Delta T

$$

Substituting:

$$
\begin{gather*}
Q_{H}=nC_{V}(T_{c}−T_{b})>0\quad ,\mathrm{Heat\ added},\ T\uparrow\\
Q_{C}=nC_{V}(T_{a}-T_{d})<0\quad ,\mathrm{Heat\ removed},\ T\downarrow
\end{gather*}
$$

The thermal efficiency $e$ is given then by: 

$$
e=\frac{Q_{H}+Q_{C}}{Q_{H}}=\frac{T_{c}−T_{b}+T_{a}−T_{d}}{T_{c}−T_{b}} 
$$
 
 

Because processes $a-b$ and $c-d$ are adiabatic, we know that: 

$$
T_{a}V_{a}^{γ−1}=T_{b}V_{b}^{γ−1}\quad \mathrm{and}\  
\quad T_{c}V_{c}^{γ−1}=T_{d}V_{d}^{γ−1}
$$
, and by remembering that $V_{a}=V_{d}=rV$ and $V_{b}=V_{c}=V$:

$$

\begin{gather*}
T_{a}(rV)^{γ−1}=T_{b}V^{γ−1}\\
T_{c}V^{γ−1}=T_{d}(rV)^{γ−1}
\end{gather*}

$$

, where $\gamma$ is the ratio of heat capacities of the gas. 
Substituting into {numref}`eq:eff_Otto`: 

$$

e=\frac{T_{d}r^{γ−1}−T_{a}r^{γ−1}+T_{a}−T_{d}}{T_{d}r^{γ−1}−T_{a}r^{γ−1}}=\frac{(T_{d}−T_{a})(r^{γ−1}−1)}{(T_{d}−T_{a})r^{γ−1}}=1−\frac{1}{r^{γ−1}}

$$

In the idealized model of the Otto cycle, it is worth noting that the thermal efficiency is always less then 1. Using as reference values $r=8$ and $\gamma=1.4$ (for air) the theoretical efficiency is $e\approx 56 \%$. In real internal-combustion engines, where the mixture does not behave as an ideal gas and when phenomena such as friction, heat dissipation and turbulence are considered, efficiency is lower (typically around $\approx 35 \% $). 

Increasing the compression ratio would result in an higher efficiency, but in reality this also increases the temperature at the end of the compression stroke. This can result in a spontaneous ignition of the fuel, which is known as detonation or pre-ignition, which can eventually damage the engine. 

 

### Diesel cycle

Diesel engine operates similarly to the gasoline engine. Its cycle is called the Diesel cycle with one one main difference that no fuel is injected in the cylinder at the beginning of the compression stroke. This difference permits the increase of the compression ratio as it is removing the risk of pre-ignition of the fuel.

The Diesel cycle is shown in {numref}`Diesel_pv` and is composed of 2 adiabatic steps, in addition to an isochoric and isobaric step. The steps are summarized below:
- **Step 0**: Intake stroke. Only air enters the cylinder (no fuel)
- **Step 1 ($a-b$)**: Compression stroke. The intake valve closes and the piston compresses the air. This is an adiabatic process, where the gas is compressed from $rV$ to $V$
- **Step 2 ($b-c$)**: Fuel ignition. This step is the main difference from the Otto cycle. Intake valve injects fuel fast enough to keep the pressure constant. Fuel ignites spontaneously (no spark plug) as high temperature is developed during adiabatic compression ($a-b$)
- **Step 3 ($c-d$)**: Power stroke. The hot mixture expands with pistons being pushed down. This is an adiabatic expansion step, in which the volume is expanded from $V$ back to $rV$.
- **Step 4 ($d-a$)**: Exhaust stroke. The exhaust valve opens and combustion residuals are expelled. The cylinder is ready for the next cycle. This is an isochoric process, where heat $Q_{C}$ id rejected to the cold reservoir.  

```{figure} figures_for_chapter_6/DieselCycle.png
:width: 60%
:name: Diesel_pv
$pV$ diagram of Diesel cycle. Red curves represent adiabatic processes, while purple and yellow ones stand for isochoric and isobaric transformation respectively.
```
The Diesel cycle enables theoretical compression ratios of 15-20 and a theoretical efficiency of $\approx 70 \% $ (using $\gamma\approx 1.4$ for air).

Summarizing:
- Otto cycle: Fuel+air before compression $\Rightarrow$ Risk of preignition before completing compression;
- Diesel cycle: compressing air before injecting fuel $\Rightarrow$ No preignition, higher compression ratios and higher thermal efficiency.

## Refrigerators and refrigeration cycle
Refrigerators can be seen as heat engines operating in reverse. It takes heat from a cold place (the inside of the refrigerator) and gives it off to a warmer place (air in the room); in order to do that the refrigerator requires a net input of mechanical work. 
With the sign conventions already introduced, for a refrigerator $Q_{C}$ is positive but both $W$ and $Q_{H}$ are negative.
Hence:

$$

|W|=-W\quad\quad |Q_{H}|=-Q_{H}

$$

[FIG NEEDED OF REFRIGERATOR DIAGRAM]

From the first law of thermodynamics:

$$

Q_{H}+Q_{C}-W=0 \Rightarrow |Q_{H}|=Q_{C}+|W|

$$

The best refrigerator is the one that is able to remove the largest amount of $Q_{C}$ with the smallest amount of work $W$. The figure of merit to classify performance is then defined $K$, i.e. coefficient of performance:

$$

K=\frac{Q_{C}}{|W|}=\frac{Q_{C}}{|Q_{H}|-Q_{C}}

$$

## Second law of thermodynamics
The experimental impossibility to build a heat engine characterized by 100% effiency is the basis of the second law of thermodynamics.
Historically different statements have been provided to explain this phenomenon:
- 1850: German scientist Rudolf Clausius formulates his statement of second law, which states:

```{admonition} Definition: Clausius statement
:class: definition
It is impossible for any process to have as its sole result the transfer of heat from a cooler to 
a hotter object.
```
This corresponds in having a 100% workless refrigerator, which from experience is not possible.

- Later, both Lord Kelvin and Max Planck formulated another statement known as Kelvin-Planck statement or heat engine statement of second law:

```{admonition} Definition: Kelvin-Planck statement
:class: definition
It is impossible to build a heat engine whose only outcome is to absorb energy in the form of heat from a single thermal reservoir and to deliver an equivalent amount of work.
```
This corresponds to a heat engine characterized by 100% thermal efficiency, which has already been shown to be not possible.

It is important to point out that neither of these impossibilities violates the first law of thermodynamics. The second law, therefore, cannot be deducted from the first but comes out by itself as a separate law of nature.

### Equivalence between Kelvin and Clausius statement
It is possible to show that the two statements of second law are completely equivalent.
To do that, let us imagine to have an "anti-Clausius" system, i.e. a workless refrigerator and to attach in series a regular heat engine, as reported in Fig.

[FIG NEEDED OF ANTICLAUSIUS+HEAT ENGINE]

Writing down first law equation for the total system:

$$

\underbrace{Q_{C}-|Q_{C}|}_{\mathrm{anti-Clausius}}+\underbrace{Q_{H}-|Q_{C}|+W}_{\mathrm{heat-engine}}=0 \Rightarrow (Q_{H}-|Q_{C}|)+W=0

$$

, which corresponds to the so-called "anti-Kelvin" system, i.e. an heat engine capable of extracting heat from a single reservoir and converting it entirely to work.

The two statements can be proven to be equivalent also starting from an anti-Kelvin engine in series with a regular refrigerator, which eventually leads to the Clausius statement.
