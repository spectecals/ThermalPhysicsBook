Considering the first law of thermodynamics applied at the case of thermodynamic cyclic processes, this means that: 

$$

U_{2}-U_{1}=0\Rightarrow Q=W,

$$

 i.e. the net heat flowing in heat engines equals the net work done.
All heat engines absorb heat $Q_{H}$ from a source at high temperature $T_{H}$, and after performing some mechanical work, discard some heat $Q_{C}$ at a lower temperature $T_{C}$.

The net heat $Q$ per cycle is: 

$$

Q=Q_{H} + Q_{C}

$$

In the particular case of the heat engine, as also shown from {numref}`Heat_engine_schematic`, $Q_{C}$ is negative since it's trasferred from the system to the cold reservoir, while $Q_{H}$ is positive. The net heat then can be also written as:

$$

Q=Q_{H}-|Q_{C}|

$$

The net work $W$ per cycle is equal to the net heat since we are dealing with a cyclic process: 

$$

W=Q_{H}+Q_{C}=Q_{H}-|Q_{C}|

$$

A way to quantify how well the engine is able to convert heat into work is by calculating the thermal efficiency $e$: 

$$

e=\frac{W}{Q_{H}} =1+\frac{Q_{C}}{Q_{H}}=1-\frac{|Q_{C}|}{Q_{H}}

$$
 
Since the heat rejected to the enviroment $Q_{C}$ in a heat engine is always negative, the efficiency of a heat engine is always less than unity. This fundamental limitation is expressed by the **Kelvin-Planck statement of the second law of thermodynamics**, which states that no heat engine can convert all the heat it receives into work. A portion of the energy must always be rejected to a lower-temperature environment. This limitation is not caused by friction or other practical imperfections. Even an ideal heat engine cannot achieve 100% efficiency. The maximum possible efficiency depends only on the temperatures of the hot and cold reservoirs, a result that will be discussed in the context of the Carnot cycle.




 
 

## Carnot cycle
According to second law of thermodynamics, a heat engine with a 100% efficiency cannot exist.
What is then the maximum conversion from heat to work reached by a heat engine?

The answer was given in 1824 by the French engineer Sadi Carnot, who 
developed a hypothetical, idealized heat engine that has the maximum possible efficiency  consistent with the second law. The cycle of this engine is called the Carnot cycle.

For maximum heat engine efficiency all processes must be reversible. This results in:
-  No temperature drops, i.e. isothermal processes at $T_{H}$ and $T_{C}$;
- When $T_{C}<T<T_{H}$, there is no heat transfer between heat engines and reservoirs, i.e. adiabatic processes occur.

```{figure} figures_for_chapter_6/CarnotFridge.png
:width: 60%
:name: Carnot_pv
$pV$ diagram of the Carnot engine cycle operating between $T_{H}$ and $T_{C}$. Blue curves correspond to isothermal processes, while red ones correspond to adiabatic transformations.
```

The Carnot cycle consists of four reversible processes: two isothermal processes and two adiabatic processes. The thermodynamic states of the Carnot cycle are illustrated in the $pV$ diagram shown in {numref}`Carnot_pv`. The cycle proceeds through the following four reversible processes:

1) **Process $a-b$: Reversible isothermal expansion**. The working gas is palced in contact with a hot reservoir at temperature $T_{H}$. The gas expands slowly while absorbing heat $Q_{H}$ from the reservoir. Because the temperature difference between the gas and the reservoir is infinitesimally small, the process is reversible and occurs at constant temperature.
2)  **Process $b-c$: Reversible adiabatic expansion**. The thermal contact with the reservoir is removed and the cylinder is thermally insulated. The gas continues to expand while performing work on the surroundings. Since no heat is exchanged during this process, the temperature of the gas decreases from $T_{H}$ to $T_{C}$.
3)  **Process $c-d$: Reversible isothermal compression**. The gas is brought into contact with the cold reservoir at temperature $T_{C}$. The surroundings compress the gas while heat $Q_{C}$ is rejected to the reservoir. The compression occurs at constant temperature.
4)  **Process $d-a$: Reversible adiabatic compression**. The gas is again thermally isolated and compressed further until its temperature increases from $T_{C}$ and $T_{H}$, returning the system to its initial state and completing the cycle.

On the $pV$ diagram the area enclosed by the cycle represents the net work delivered by the heat engine. The upper portion of the cycle corresponds to expansion processes where work is produced, while the lower portion corresponds to compression processes where work is done on the working fluid.

### The reversed Carnot cycle

Because all processes of the Carnot cycle are reversible, the entire cycle can be reversed. When the direction of the cycle is reversed, the device operates as a refriegerator or heat pump. In this case, heat $Q_{C}$ is extracted from the cold reservoir, heat $Q_{H}$ is rejected to the hot reservoir, and a net work input $W$ is required to drive the cycle. This reversed operation corresponds to the cycle shown in {numref}`Carnot_fridge_scheme` and {numref}`Carnot_fridge_pv`.

```{figure} figures_for_chapter_6/RefrigeratorSchematic.png
:width: 60%
:name: Carnot_fridge_scheme
General schematic of a Carnot fridge operating between $T_{H}$ and $T_{C}$.
```

Since the work produced by the super-Carnot engine matches the work required by the fridge to work, the resulting system will consist of a transfer of heat from the cold reservoir to the hot one without any work required. 

Moreover, from the definition of efficiency of the Carnot fridge given above, it is possible to observe:

$$

e_{CF}=\frac{Q_{C}}{W},\quad e_{CF}=\frac{Q_{C}^{CF}}{W(1+x)}\\
\Rightarrow Q_{C}^{CF}=Q_{C}(1+x)=\left(Q_{H}-W\right)(1+x)

$$

and then the heat transferred to the hot reservoir by the Carnot fridge will be:

$$

Q_{H}^{CF}=Q_{C}(1+x)+W(1+x)=\underbrace{(Q_{C}+W)}_{Q_{H}}(1+x)

$$

Let us express now also the total amount of heat rejected to the hot reservoir:

$$

Q_{H}-Q_{H}^{CF}=Q_{H}-Q_{H}(1+x)=-Q_{H}x,

$$

while the total heat extracted from the cold reservoir to the system will correspond

$$

-Q_{C}^{SC}+Q_{C}^{CF}=\underbrace{-Q_{H}+W(1+x)}_{-Q_{C}^{SC}}+\underbrace{\left(Q_{H}-W\right)\left(1+x\right)}_{Q_{C}^{CF}}=Q_{H}x

$$

Calculating the variation of entropy:

$$

\Delta S=\Delta S_{H}+\Delta S_{C}=+\frac{Q_{H}x}{T_{H}}-\frac{Q_{H}x}{T_{C}}=xQ_{H}\left(\frac{1}{T_{H}}-\frac{1}{T_{C}}\right)<0,

$$

which is negative since $T_{H}>T_{C}$. This clearly violates the second law of thermodynamics and then it is false to assume a super Carnot engine to exist with $x>0$. To satisfy second law, $x\leq 0$, which means the Carnot engine is the most efficient heat engine.

### Efficiency of Carnot cycle
To calculate the efficiency of the Carnot engine let's calculate:

$$

Q_{H}=W_{ab}=nRT_{H}\mathrm{ln}\frac{V_{b}}{V_{a}}>0,

$$

while:

$$

Q_{C}=W_{cd}=nRT_{C}\mathrm{ln}\frac{V_{d}}{V_{c}}<0

$$

In adiabatic processes $b-c$ and $d-a$, we know from Moduòe 4.10 that the pressur-volume relationship is:

$$

\begin{gather*}
pV^{\gamma}=\mathrm{constant},\quad \gamma=\frac{C_{p}}{C_{V}}\\
\Rightarrow p_{1}V_{1}^{\gamma}=p_{2}V_{2}^{\gamma}
\end{gather*}

$$

Substituting into the ideal gas low we have:

$$

nRT\frac{V^{\gamma}}{V}=\mathrm{constant}\\
\Rightarrow TV^{\gamma-1}=\mathrm{constant}

$$

and then for processes $b-c$ and $d-a$:

$$

\begin{gather*}
T_{H}V_{b}^{\gamma-1}=T_{C}V_{c}^{\gamma-1}\\
T_{C}V_{d}^{\gamma-1}=T_{H}V_{a}^{\gamma-1}
\end{gather*},

$$

which then results:

$$

\frac{V_{b}}{V_{a}}=\frac{V_{c}}{V_{d}}

$$

The thermal efficiency of the cycle is then:

$$

e=1+\frac{Q_{C}}{Q_{H}}=1-\frac{T_{C}}{T_{H}}\underbrace{\frac{\mathrm{ln}\frac{V_{c}}{V_{d}}}{\mathrm{ln}\frac{V_{b}}{V_{a}}}}_{=1}=1-\frac{T_{C}}{T_{H}}

$$

Equivalently, the coefficient of performance ($COP$) of the Carnot refrigerator can be proven to be:

$$

COP=\frac{Q_{C}}{|W|}=\frac{T_{C}}{T_{H}-T_{C}}

$$
, where now $W$ is negative (input work required to drive the cycle).

### Carnot Principles

Two fundamental conclusions follow from the second law of thermodynamics. First, the efficiency of any real (irreversible) heat engine is always lower than that of a reversible engine operating between the same two temperature reservoirs. Second, all reversible heat engines operating between the same temperature have exactly the same efficiency, regardless of the working fluid or the specific details of the cycle. These results, known as the Carnot principles, establish the Carnot cycle as the benchmark for the maximum achievable efficiency of heat engines.
The Carnot cycle is not a practical engine cycle and cannot be implemented exactly in real machines. Nevertheless, it plays a fundamental role in thermodynamics beacuse it establishes the maximum possible efficiency that any heat engine can achieve between two given temperature reservoirs. No real engine operating between the same temperatures can be more efficient than a Carnot engine.

### Otto cycle
The Otto cycle is composed of 4 main processes: two adiabatic  and two isochoric processes summarized in  {numref}`Otto_pv`, and outlined below:
- **Step 0**: Intake stroke. Here the piston moves down, allowing gasoline-air mixture to enter the cylinder via the intake valve.
- **Step 1 ($a-b$)**: Compression stroke. The intake valve closes and the piston compresses the misture (adiabatic process).
- **Step 2 ($b-c$)**: Ignition. The spark plug ignites the mixture.$Q_{H}$ is added to the system by the burnt fuel. This is an isochoric step.
- **Step 3 ($c-d$)**: Power stroke. The hot mixture expands with pistons being pushed down. This is an adiabatic expansion step, in which the volume is expanded from $V$ back to $rV$.
- **Step 4 ($d-a$)**: Exhaust stroke. The exhaust valve opens and combustion residuals are expelled. The cylinder is ready for the next cycle. This is an isochoric process, where heat $Q_{C}$ id rejected to the cold reservoir. 
 

```{figure} figures_for_chapter_6/OttoCycle.png
:width: 60%
:name: Otto_pv
$pV$ diagram of the Otto cycle. Red curves represent adiabatic processes, while purple curves represent the isochoric processes.
```


Let us now calculate the efficiency of the Otto cycle.
Let us start from the ioschoric processes $b-c$ and $d-a$. An isochoric process is a process at constant volume. Since the volume is not changed, no work is done, $W=0$, as:

$$

dW=pdV

$$

From the first law of thermodynamics we know that $\Delta U=Q-W=Q$. $Q$ is then given by:

$$

Q=nC_{V}\Delta T

$$

$Q_{H}$ and $Q_{C}$ can then be written as:

$$
\begin{gather*}
Q_{H}=nC_{V}(T_{c}−T_{b})>0\quad ,\mathrm{Heat\ added},\ T_{c}>T_{b}\\
Q_{C}=nC_{V}(T_{a}-T_{d})<0\quad ,\mathrm{Heat\ removed},\ T_{a}<T_{b}
\end{gather*}
$$

The thermal efficiency $e$ of the Otto cycle is given by: 

$$
e=\frac{Q_{H}+Q_{C}}{Q_{H}}=\frac{T_{c}−T_{b}+T_{a}−T_{d}}{T_{c}−T_{b}} 
$$
 
 

Because processes $a-b$ and $c-d$ are adiabatic, we know from Modules 4.10 and 6.2 that: 

$$
T_{a}V_{a}^{γ−1}=T_{b}V_{b}^{γ−1}\quad \mathrm{and}\  
\quad T_{c}V_{c}^{γ−1}=T_{d}V_{d}^{γ−1}
$$
, and by remembering that $V_{a}=V_{d}=rV$ and $V_{b}=V_{c}=V$:

$$

\begin{gather*}
T_{a}(rV)^{γ−1}=T_{b}V^{γ−1}\\
T_{c}V^{γ−1}=T_{d}(rV)^{γ−1}
\end{gather*}

$$

, where $\gamma$ is the ratio of heat capacities of the gas. 
Substituting into {numref}`eq:eff_Otto`: 

$$

e=\frac{T_{d}r^{γ−1}−T_{a}r^{γ−1}+T_{a}−T_{d}}{T_{d}r^{γ−1}−T_{a}r^{γ−1}}=\frac{(T_{d}−T_{a})(r^{γ−1}−1)}{(T_{d}−T_{a})r^{γ−1}}=1−\frac{1}{r^{γ−1}}

$$

In the idealized model of the Otto cycle, it is worth noting that the thermal efficiency is always less then 1. Using as reference values $r=8$ and $\gamma=1.4$ (for air) the theoretical efficiency is $e\approx 56 \%$. In real internal-combustion engines, where the mixture does not behave as an ideal gas and when phenomena such as friction, heat dissipation and turbulence are considered, efficiency is lower (typically around $\approx 35 \% $). 

Increasing the compression ratio would result in an higher efficiency, but in reality this also increases the temperature at the end of the compression stroke. This can result in a spontaneous ignition of the fuel, which is known as detonation or pre-ignition, which can eventually damage the engine. 

 

### Diesel cycle

The Diesel engine operates similarly to the gasoline engine. Its cycle is called the Diesel cycle with one one main difference that no fuel is injected in the cylinder at the beginning of the compression stroke compared to the Otto cycle. This difference permits the increase of the compression ratio as it removes the risk of pre-ignition of the fuel.

The Diesel cycle is shown in {numref}`Diesel_pv` and is composed of two adiabatic steps, in addition to an isochoric and isobaric step. The steps are summarized below:
- **Step 0**: Intake stroke. Only air enters the cylinder (no fuel).
- **Step 1 ($a-b$)**: Compression stroke. The intake valve closes and the piston compresses the air. This is an adiabatic process, where the gas is compressed from $rV$ to $V$.
- **Step 2 ($b-c$)**: Fuel ignition. This step is the main difference from the Otto cycle. Intake valve injects fuel fast enough to keep the pressure constant. Fuel ignites spontaneously (no spark plug) as high temperature is developed during adiabatic compression ($a-b$).
- **Step 3 ($c-d$)**: Power stroke. The hot mixture expands with pistons being pushed down. This is an adiabatic expansion step, in which the volume is expanded from $V$ back to $rV$.
- **Step 4 ($d-a$)**: Exhaust stroke. The exhaust valve opens and combustion residuals are expelled. The cylinder is ready for the next cycle. This is an isochoric process, where heat $Q_{C}$ is rejected to the cold reservoir.  

```{figure} figures_for_chapter_6/DieselCycle.png
:width: 60%
:name: Diesel_pv
$pV$ diagram of the Diesel cycle. The red curves represent adiabatic processes, while the purple and yellow curves represent isochoric and isobaric processes respectively.
```
The Diesel cycle enables theoretical compression ratios of 15-20 and a corresponding theoretical efficiency of $\approx 70 \% $ (using $\gamma\approx 1.4$ for air).

In summary, the key distinction between the Otto and Diesel cycle lies in when fuel is introduced with respect to compression stroke, and this fundamentally shapes their performance limits. The Otto cycle's pre-mixed fuel-air mixture limits compression due to risk of premature ignition cappin its efficiency. In contrast, the Diesel cycle compresses only air and injects fuel afterwards, eliminating the constraint and enabling significantly higher compression ratios. This make the Diesel engine more efficient.

## Refrigerators and refrigeration cycle

In nature, heat is transferred spontaneously from regions of higher temperature to regions of lower temperature. For example, a hot object placed in a cooler environment will naturally lose heat until thermal equilibrium is reached.
The reverse process, however, cannot occur spontaneously. Transferring heat from a colder region to a warmer one requires the use of special devices known as **refrigerators**. Like heat engines, refrigerators operate on thermodynamic cycles and use a working fluid called a **refrigerant**.

With the sign conventions already introduced, for a refrigerator $Q_{C}$ is positive but both $W$ and $Q_{H}$ are negative.
Hence:

$$

|W|=-W\quad\quad |Q_{H}|=-Q_{H}

$$

[FIG NEEDED OF REFRIGERATOR DIAGRAM]

From the first law of thermodynamics:

$$

Q_{H}+Q_{C}-W=0 \Rightarrow |Q_{H}|=Q_{C}+|W|

$$

The best refrigerator is the one that is able to remove the largest amount of $Q_{C}$ with the smallest amount of work $W$. 
The effectiveness of a refrigerator is measured by the **coefficient of performance** ($COP$), which quantifies how much heat can be removed from the cold region per unit of work supplied.

For a refrigerator, the desired effect is the removal of heat $Q_{C}$ from the refrigerated space, while the required input is the work $W$ supplied to the compressor. The coefficient of performance of a refrigerator is therefore defined as:

$$

COP=\frac{Q_{C}}{|W|}=\frac{Q_{C}}{|Q_{H}|-Q_{C}}

$$

Unlike the thermal efficiency of a heat engine, the coefficient of performance of a refrigerator can be greater than unity. This simply means that the amount of heat removed from the refrigerated space can be larger than the mechanical work supplied to the compressor.


## Rankine cycle
In Modulke 6.3 gas power cycles were introduced, in which the working fluid remains in the gaseous phase throughout the entire cycle. In this chapter, we turn to vapour power cycles, where the working fluid is alternately vaporized and condensed. These cycles are widely used in large-scale power generation and can also be coupled with process heating in applications known as cogeneration.
The increasing demand for higher thermal efficiencies in power generation has driven the development of improved vapour power cycle configurations. These include modifications such as reheat, regenerative feedwater heating, and combined gas-vapour cycles, all aimed at extracting more useful work from a given heat input.

Among the various working fluids available, water (steam) is the most commonly used due to its low cost, availability, and high latent heat of vaporization. Although steam power plants may differ in their heat source – such as coal, nuclear, or natural gas – the fundamental thermodynamic processes remain largely the same.
To understand the design and performance of these systems, it is useful to first consider the theoretical limits of efficiency and the practical constraints that arise in real cycles. This leads to the development of the Rankine cycle, which forms the basis of modern steam power plants.

Before introducing the Rankine cycle, it is useful to recall the Carnot cycle, which represents the maximum theoretical efficiency achievable between two temperature limits. However, despite its theoretical importance, the Carnot cycle cannot be implemented in real systems due to several practical limitations:

- Isothermal heat addition and rejection require infinitely slow heat transfer: Maintaining constant temperature during heat transfer requires an infinitesimally small temperature difference between the working fluid and the heat source/sink. On a pV diagram, these processes appear as smooth curves, but in reality achieving them would require unrealistically long times and extremely large heat exchangers. 
- Compression of a wet vapour mixture is mechanically impractical: During the Carnot cycle, compression occurs in the two-phase region. Compressing a vapour can induce condensation, forming liquid droplets within the flow (recall Module 3.4). These droplets can damage compressors or turbine blades and lead to severe efficiency losses, making the process mechanically unsafe and energetically inefficient.

While Otto and Diesel cycles describe internal combustion engines, steam power plants operate on a fundamentally different principle based on external heat addition, where the working fluid is separate from the combustion process.

The Rankine cycle avoids the key limitations of the Carnot cycle by:
- Replacing isothermal heat transfer with constant-pressure processes 
- Ensuring that compression occurs only in the liquid phase 
- Allowing expansion to occur primarily in the vapour phase 

These modifications make the cycle feasible for power generation systems. The physical layout of the Rankine cycle is provided in {numref}`Rankine_schematics` and is composed of a pump, boiler, turbine and condensor.

[FIG NEEDED OF RANKINE SCHEMATICS]

The Rankine Cycle stages (refer to $$pV$$ diagram, {numref}`Rankine_diagram`) can be clearly identified on the $pV$ diagram:

- 1 → 2: Pump (liquid compression (isoentropic)). Water enters as a saturated liquid from the condenser. It is compressed to a higher pressure by the pump. On the $pV$ diagram, this appears as a nearly vertical line (very small volume change). Because the fluid is liquid, no phase change occurs, and the work required is small.  

- 2 → 3: Boiler (constant-pressure heat addition)
The high pressure water water is heated to high pressure steam at constant pressure:
    - First, temperature rises (compressed liquid region) 
    - Then phase change occurs (liquid → vapour) 
    - Optionally, the vapour is superheated ( shown in {numref}`super_heated_rankine`)
On the $pV$ diagram, this corresponds to a large increase in specific volume as the water evaporates. Note that some common heat sources for the boiler include: burning fossil fuels (e.g. coal, natural gas), nuclear energy (heat fromn nuclear fission), solar energy, geothermal (heat from hot fluids in the ground).

- 3 → 4: Turbine (adiabatic – more accurately it is isoentropic – expansion)
High-pressure steam expands through the turbine, producing mechanical work, which.can be used for example to spin a generator.
Pressure and temperature decrease, and the fluid may enter the wet region (vapour–liquid mixture). On the $pV$ diagram, this is a downward-sloping expansion curve. 

- 4 → 1: Condenser (constant-pressure heat rejection) 
Heat is rejected at constant pressure, and the steam condenses back into liquid. On the $pV$ diagram, the volume decreases significantly as the fluid returns to the saturated liquid state. 








## Nuclear power
Thermodynamic principles also apply to systems that exploit nuclear reactions to generate electricity. Energy can exist in many forms—thermal, mechanical,kinetic, potential, chemical, electrical, and nuclear—and their sum constitutes the total energy of a system. In thermodynamic analysis we are usually interested in the change of energy rather than its absolute value. In most engineering
applications, the energy extracted from a source is converted into thermal energy and subsequently into mechanical work by means of a heat engine.
Nuclear power plants operate according to the same thermodynamic principle as conventional thermal power plants. The key difference lies in the source of heat. Instead of burning fossil fuels, nuclear reactors generate heat through nuclear reactions occurring inside the reactor core. The thermal energy produced by these reactions heats water to produce steam, which expands through a turbine connected to an electrical generator, thereby converting thermal energy into mechanical and then electrical energy.

{numref}`NUCLEAR_POWER_PLANT` illustrates the basic layout of a nuclear power station. Heat generated inside the reactor building is transferred to water in a boiler system, producing high-pressure steam. The steam drives a turbine connected to a generator, which produces electricity that is transmitted to the power grid. A cooling system removes excess heat and condenses the steam so that the cycle can continue.

Among the possible reactions, fission of unstable nuclei are the ones exploited to retrieve the needed energy to sustain the nuclear power plant.
Inside the reactor the reaction occurring is:

$$

_{92}^{235}\mathrm{U}+n\rightarrow\  _{56}^{142}\mathrm{Ba}+\ _{36}^{91}\mathrm{Kr}+3n+E

$$

, where top-left numbers correspond to mass number (protons+neutrons), the bottom-left one to the atomic number, and $n$ stands for neutrons and $E$ for the released energy due to the difference in binding energies of atoms. In fact, it holds that:

$$
    E=mc^{2}
$$
, where $m$ is the mass of the atom and $c$ is the speed of light.

```{admonition} Example
:class: example

Uranium $_{92}^{238}\mathrm{U}$ decays into Thorium  $_{90}^{234}\mathrm{Th}$ and an $\alpha$ particle. Calculate how much energy is released in the process.

The mass of $_{92}^{238}\mathrm{U}=238.0508\mathrm{u}$, mass of $_{90}^{234}\mathrm{Th}=234.0426\mathrm{u}$ and mass of $_{2}^{4}\alpha=4.0026\mathrm{u}$.


Remembering that $1\mathrm{u}=1.66\times 10^{-27} \mathrm{kg} $:

$$

E=(238.0508-234.0426-4.0026)\cdot 1.66\cdot 10^{-27} \mathrm{kg}\cdot c^{2}=8.37\cdot 10^{-13}\ J

$$

```
Coming back to the nuclear reaction describing uranium decay, we observe the presence of 3 neutrons as reaction products, which can collide with other uranium atoms, thus giving rise to a chain reaction.

[FIG NEEDED OF CHAIN REACTION DECAY IN JL NOTES]

In order to maximize the efficiency of the reaction, neutrons need to be slowed down. To achieve that, heavy water is used ($\mathrm{D}_{2}\mathrm{O}$) is used.
To control the chain reaction, rods made of boron are implemented (they absorb neutrons): when rods are inserted in the reactor, the reaction is slowed down, while it speeds up when rods are removed.

Another possible source of nuclear energy is **fusion**, in which tow light nuclei combine to form a heavier nucleus while releasing energy. An example is the fusion reaction of two deuterium nuclei (deuterium is an isotope of hydrogen):

$$

    ^{2}_{1}\mathrm{H}+^{2}_{1}\mathrm{H}\rightarrow\ ^{3}_{2}\mathrm{He}+n

$$

Fusion reactions are responsible for the enormous energy output of the Sunand other stars. However, achieving controlled fusion on Earth is extremely challenging because the positevly charged nuclei strongly repel each other through Coulomb forces. To overcome this repulsion, the reacting particles must reach temperatures on the order of $10^{8}$ K, conditions typically found only in stellar interiors or nuclear explosions.

```{admonition} A Car Powered by Nuclear Fuel
:class: example
**Question** An average car consumes about 5 L of gasoline per day. Assume that gasoline has average density

$$

\rho = 0.75\ \mathrm{kg}\ \mathrm{L}^{-1}

$$

and lower heating value

$$

H = 44000\ \mathrm{kJ}\ \mathrm{kg}^{−1}

$$

Suppose that all practical difficulties associated with radioactivity and waste disposal are ignored, and that a car is instead powered by uranium-235. If the car is equipped with 0.1 kg of uranium-235, determine whether the car would ever need refueling under average driving conditions.

**Solution**. We first calculate the mass of gasoline consumed by the car in one day. Using the given density and daily fuel consumption,

$$

m_{gasoline} = \rho V = (0.75\ \mathrm{kg}\ \mathrm{L}^{−1})(5\ \mathrm{L}/\mathrm{day}) = 3.75\ \mathrm{kg}/\mathrm{day}

$$

The corresponding energy supplied to the car per day is therefore

$$

E_{day} = m_{gasoline}H = (3.75\ \mathrm{kg/day})(44000\ \mathrm{kJ}\ \mathrm{kg}^{−1}) = 165000\ \mathrm{kJ/day}

$$

Next, we calculate the total energy available from the nuclear fuel. The complete fission of 1 kg of uranium-235 releases approximately $6.73\times 10^{10}\ \mathrm{kJ/kg}$.

Therefore, for $0.1$ kg of uranium-235, the total energy available is:

$$
E_{nuclear} = (6.73\times 10^{10}\ \mathrm{kJ/kg})(0.1\ \mathrm{kg}) = 6.73\times 10^{9}\ \mathrm{kJ}

$$

The number of days for which this amount of nuclear fuel could supply the car is

$$
N =\frac{E_{nuclear}}{Eday}=\frac{6.73 \times 10^{9}\ \mathrm{kJ}}{165000\ \mathrm{kJ/day}}\approx 40,790\ \mathrm{days}.

$$

This corresponds to approximately $\approx 112$ years. Therefore, from a purely energetic point of view, the car would not need
refueling during its lifetime.

**Equivalent power consumption during the same period**
It is useful to compare this result with the amount of gasoline that would be required to provide the same total energy. Since the car consumes $3.75$ kg of gasoline per day, the total gasoline mass needed over $40790$ days would be:

$$

m_{total} = (3.75\ \mathrm{kg/day})(40790\ \mathrm{days})\approx 1.53\times 10^{5}\ \mathrm{kg}

$$

Thus, about $150$ tons of gasoline would be required to supply the same amount of energy as $0.1$ kg of uranium-235.
This comparison illustrates the extremely high energy density of nuclear fuel
compared with conventional chemical fuels.

```
