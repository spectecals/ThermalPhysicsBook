```

```{admonition} Learning outcomes:
:class: outcomes

- Explain the role and importance of thermodynamic systems and processes.
- Learn about different thermodynamic cycles.
- Calculate thermal effiency of a heat engine.
- Identify the difference between heat engines and refrigerators.
- Calculate the efficiency of the corresponding Carnot cycle.

## Introduction and therminology
Any device that transforms heat partially into work or mechanical energy is called heat engine.  Independently from the type of engine, the working principle includes a quantity of matter, here defined as working substance, that undergoes through different thermodynamic processes e.g. expansion, compression, inflow or outflow of heat etc... 

It is also possible to distinguish heat engines as open processes or cyclic processes. One example of the former, that will be explored in \ref{ref: internal combustion} is the case of internal-combustion engines. In fact, these engines exploit a mixture of air and fuel, which eventually is discarded. 
From now on heat engines will be described only as cyclic processes (it is possible to approximate open processes as cyclic). 

Considering the first law of thermodynamics applied at the case of thermodynamic cyclic processes, this means that: 

$$
U_{2}-U_{1}=0\Rightarrow Q=W
$$

, i.e. the net heat flowing in heat engines equals the net work done.
All heat engines absorb heat from a source at high temperature, and after performing some mechanical work, discard some heat at a lower temperature.

When studying heat engines, it is useful as a zeroth-order approximation to define two objects with which the working substance can interact: 
- Hot reservoir, which represents the heat source, and provides the working substance with large amounts of heat without reducing its temperature $T_{H}$. 
- Cold reservoir, able to absorb large quantities of heat discarded by the working substance without increasing its temperature $T_{C}$.

That being said, it is possible to represent the energy transformation in a heat engine through an energy-flow diagram, in which the engine is represented by a circle, surrounded at the top by the hot reservoir, and the bottom by the cold one. 
The amount of heat provided by the hot reservoir to the engine $Q_{H}$ is represented by a pipeline connecting the two elements, whose width is proportional to its magnitude. The same notation is used to describe the amount of heat discarded from the engine to the cold reservoir $|Q_{C}|$.
An additional branch coming out of the engine represent the amount of heat that has been converted to mechanical work $W$.


 
 

The net heat $Q$ per cycle is: 

$$
Q=Q_{H} +Q_{C}=Q_{H}−|Q_{C}|
$$

The net work $W$ per cycle is equal to the net heat since we are dealing with a cyclic process: 

$$
W=Q_{H}−|Q_{C}|
$$

The maximum amount of work would be achieved if it was possible to convert all $Q_{H}$ and by having $Q_{C}=0$. From experience this is not possible, as there will always be some heat wasted. 

A way to quantify how well the engine is able to convert heat into work is by calculating the thermal efficiency $e$: 

$$

e=\frac{W}{Q_{H}} =1+\frac{Q_{C}}{Q_{H}}=1−\frac{Q_{C}}{Q_{H}}∣

$$
 
 

## Internal combustion engine
An internal combustion engine is a particular type of heat engine in which energy is retrieved by means of the combustion occurring inside the working fluid (which is typically a mixture of air and fuel). 

In fig. \ref{} it is reported the working principle of one type of gasoline engine. 

[FIG NEEDED] 
### Otto cycle
Considering again fig. \ref{}, let's analyze the five steps:
- Compression stroke: the piston compresesses the mixture (adiabatic compression from $rV$ to $V$, with $r$ compression ratio);
- Ignition: the mixture is then ignited (heat $Q_{H}$ is absorbed from the hot reservoir at constant volume). 
- Power stroke: the heated gas expands (adiabatic expansion from $V$ to $rV$) 
- Exhaust stroke: combustion products are expelled through the opening of exhaust valves (exchange of heat $|Q_{C}|$ to the cold reservoir at constant volume). 

This idealized model of the thermodynamic processes in a gasoline engine is called Otto cycle. 

 





[FIG NEEDED OF OTTO CYCLE] 

It is possible to calculate the efficiency of this idealized cycle. 

Starting from processes bc and da which are constant-volume: 

$$
\begin{gather*}
Q_{H}=nC_{V}(T_{c}−T_{b})>0\\
Q_{C}=nC_{V}(T_{a}−T_{d})<0
\end{gather*}
$$

The thermal efficiency $e$ is given then by: 

$$
e=\frac{Q_{H}+Q_{C}}{Q_{H}}=\frac{T_{c}−T_{b}+T_{a}−T_{d}}{T_{c}−T_{b}}
$$
 
 

By making use of the fact that processes ab and cd are adiabatic: 

$$
T_{a}V_{a}^{γ−1}=T_{b}V_{b}^{γ−1}\quad \mathrm{and}\  
\quad T_{c}V_{c}^{γ−1}=T_{d}V_{d}^{γ−1}
$$
, and by remembering that $V_{a}=V_{d}=rV$ and $V_{b}=V_{c}=V$:

$$

\begin{gather*}
T_{a}(rV)^{γ−1}=T_{b}V^{γ−1}\\
T_{c}V^{γ−1}=T_{d}(rV)^{γ−1}
\end{gather*}

$$

, where $\gamma$ is the ratio of heat capacities of the gas. 
Substituting into Eq. \ref{}: 

$$

e=\frac{T_{d}r^{γ−1}−T_{a}r^{γ−1}+T_{a}−T_{d}}{T_{d}r^{γ−1}−T_{a}r^{γ−1}}=\frac{(T_{d}−T_{a})(r^{γ−1}−1)}{(T_{d}−T_{a})r^{γ−1}}=1−\frac{1}{r^{γ−1}}

$$

Even in the case of an idealized model as Otto cycle, it is worth observing that the thermal efficiency is always less then 1. Using as reference values $r=8$ and $\gamma=1.4$ the theoretical efficiency is $e[\%]\approx 56 \%$: in real internal-combustion engines, where the mixture does not behave as an ideal gas and phenomena such as friction, heat dissipation and turbulence are considered, efficiency is much lower (typically around 35 %). 

Increasing the compression ratio would result in an higher efficiency, but in reality this also increases the temperature at the end of the compression stroke. This can result in a spontaneous ignition of the fuel, which is known as detonation or pre-ignition, which eventually damages mechanically the engine. 

 

### Diesel cycle

The Diesel engine represents an alternative to the gasoline one, whose usefulness comes from the possibility to increase the compression ratio without risking pre-ignition of the fuel (it is injected in the cylinder after the compression stroke, right before the power stroke). 

[FIG NEEDED WITH DIESEL PV DIAGRAM] 

In fig. \ref{}  it is reported the pV diagram of the Diesel cycle. Starting at point a, air (no fuel injected in this phase) is compressed adiabatically to point b. Differently from Otto cycle, from state b to c heat is given by the hot reservoir at constant pressure.  Then thermodynamic processes cd  and da follow the same behaviour as in gasoline cycle. 

 

Since the value of the compression ratio can rise up to 20 (15-20 typical values), the theoretical efficiency of the idealized Diesel cycle is about 70%. 

## Refrigerators and refrigeration cycle
Refrigerators can be seen as heat engines operating in reverse. It takes heat from a cold place (the inside of the refrigerator) and gives it off to a warmer place (air in the room); in order to do that the refrigerator requires a net input of mechanical work. 
With the sign conventions already introduced, for a refrigerator $Q_{C}$ is positive but both $W$ and $Q_{H}$ are negative.
Hence:

$$

|W|=-W\quad\quad |Q_{H}|=-Q_{H}

$$

[FIG NEEDED OF REFRIGERATOR DIAGRAM]

From the first law of thermodynamics:

$$

Q_{H}+Q_{C}-W=0 \Rightarrow |Q_{H}|=Q_{C}+|W|

$$

The best refrigerator is the one that is able to remove the largest amount of $Q_{C}$ with the smallest amount of work $W$. The figure of merit to classify performance is then defined $K$, i.e. coefficient of performance:

$$

K=\frac{Q_{C}}{|W|}=\frac{Q_{C}}{|Q_{H}|-Q_{C}}

$$

[EXAMPLE ON JL NOTES, SIMPLE CALCULATION]

Let us look now in detail the thermodynamic transformations the working substance undergoes in a refriegeration cycle:

[FIG NEEDED OF PV DIAGRAM OF REFRIGERATION CYCLE FROM JL NOTES]

Looking at the $pV$ diagram:

1. Adiabatic expansion from the liquid substance to vapour+liquid phase
2. Isobaric expansion ending in a high $T$, low $p$ and high $V$ and absorbing heat from inside the refrigerator
3. Adiabatic compression
4. Isobaric compression

## Second law of thermodynamics
The experimental impossibility to build a heat engine characterized by 100% effiency is the basis of the second law of thermodynamics.
Historically different statements have been provided to explain this phenomenon:
- 1850: German scientist Rudolf Clausius formulates his statement of second law, which states:

```{admonition} Definition: Clausius statement
:class: definition
It is impossible for any process to have as its sole result the transfer of heat from a cooler to 
a hotter object.
```
This corresponds in having a 100% workless refrigerator, which from experience is not possible.

- Later, both Lord Kelvin and Max Planck formulated another statement known as Kelvin-Planck statement or heat engine statement of second law:

```{admonition} Definition: Kelvin-Planck statement
:class: definition
It is impossible to build a heat engine whose only outcome is to absorb energy in the form of heat from a single thermal reservoir and to deliver an equivalent amount of work.
```
This corresponds to a heat engine characterized by 100% thermal efficiency, which has already been shown to be not possible.

It is important to point out that neither of these impossibilities violates the first law of thermodynamics. The second law, therefore, cannot be deducted from the first but comes out by itself as a separate law of nature.

### Equivalence between Kelvin and Clausius statement
It is possible to show that the two statements of second law are completely equivalent.
To do that, let us imagine to have an "anti-Clausius" system, i.e. a workless refrigerator and to attach in series a regular heat engine, as reported in Fig.

[FIG NEEDED OF ANTICLAUSIUS+HEAT ENGINE]

Writing down first law equation for the total system:

$$

\underbrace{Q_{C}-|Q_{C}|}_{\mathrm{anti-Clausius}}+\underbrace{Q_{H}-|Q_{C}|+W}_{\mathrm{heat-engine}}=0 \Rightarrow (Q_{H}-|Q_{C}|)+W=0

$$

, which corresponds to the so-called "anti-Kelvin" system, i.e. an heat engine capable of extracting heat from a single reservoir and converting it entirely to work.

The two statements can be proven to be equivalent also starting from an anti-Kelvin engine in series with a regular refrigerator, which eventually leads to the Clausius statement.



## Carnot cycle
According to second law, it cannot exist a heat engine with a 100% efficiency.

In an ideal case, what is the maximum efficiency that can be reached?

The answer was given in 1824 by the French engineer Sadi Carnot, who 
developed a hypothetical, idealized heat engine that has the maximum possible efficiency 
consistent with the second law. The cycle of this engine is called the Carnot cycle.

For maximum heat engine efficiency we must avoid all irreversible processes. Therefore, considering the general working principle of an heat engine:
- During heat transfers there must not be temperature drops, since heat flow at different temperatures is an irreversible process. This means that isothermal transformation must occur
- During any other process involving temperature changes from $T_{C}$ and $T_{H}$, there must not be any heat transfer, thus imposing them to be adiabatic.

[FIG NEEDED OF CARNOT CYCLE PV DIAGRAM]

In fig. it is reported the $pV$ diagram of Carnot cycle:
- Isothermal expansion ($ab$) at temperature $T_{H}$ absorbing heat $Q_{H}$;
- Adiabatic expansion ($bc$) from $T_{H}$ to $T_{C}$;
- Isothermal compression ($cd$) at temperature $T_{C}$ rejecting heat $|Q_{C}|$;
- Adiabatic compression ($da$) from $T_{C}$ to $T_{H}$

### Efficiency of Carnot cycle
To calculate the efficiency let's calculate:

$$

Q_{H}=W_{ab}=nRT_{H}\mathrm{ln}\frac{V_{b}}{V_{a}}

$$
, while:

$$

Q_{C}=W_{cd}=nRT_{C}\mathrm{ln}\frac{V_{d}}{V_{c}}=-nRT_{C}\mathrm{ln}\frac{V_{c}}{V_{d}}

$$

In adiabatic processes $bc$ and $da$ it holds:

$$
\begin{gather*}
T_{H}V{b}^{\gamma-1}=T_{C}V_{c}^{\gamma-1}\\
T_{C}V_{d}^{\gamma-1}=T_{H}V_{a}^{\gamma-1}
\end{gather*}
$$

, which then results:

$$

\frac{V_{b}}{V_{a}}=\frac{V_{c}}{V_{d}}

$$

The thermal efficiency of the cycle is then:

$$

e=1+\frac{Q_{C}}{Q_{H}}=1-\frac{T_{C}}{T_{H}}\underbrace{\frac{\mathrm{ln}\frac{V_{b}}{V_{a}}}{\mathrm{ln}\frac{V_{c}}{V_{d}}}}_{=1}=1-\frac{T_{C}}{T_{H}}

$$




## Rankine cycle
In real life, trying to reproduce Carnot cycle is practically impossible because of:

- Isothermal heat addition

- Isothermal heat rejection

- Compression of a wet-vapour mixture 

In modern heat engines, and in particular in steam-based power plants, the most implemented thermodynamic cycle is the so-called Rankine cycle.
Differently from Carnot cycle, Rankine exploits adiabatic expansion(compressions) of purely gaseous(liquid) phases of the working substance, which is far easier to achieve mechanically and energetically.

[FIG NEEDED OF RANKINE CYCLE pV diagram+ components scheme]

Here in Fig. is a sketch on the $pV$ diagram of the thermodynamic cycle. Basically:

1) 

2) 

3) 

4) 





### Nuclear power
