1) A heat engine works on the Carnot Cycle, taking heat from burning fuel at 600 K at the rate of 1000 W. The heat engine rejects the waste heat at a temperature of 300 K. What is the efficiency of the engine and the power output.

The efficiency of the Carnot cycle is equal to:

$$

e_{C}=\frac{T_{H}-T_{C}}{T_{H}}=\frac{600\ \mathrm{K}-300\ \mathrm{K}}{600\ \mathrm{K}}=0.5

$$

Moreover, the efficiency of an heat engine is also defined as the ratio between the output work $W$ and the heat extracted from the hot reservoir $Q_{H}$:

$$

e_{C}=\frac{W}{Q_{H}}\Rightarrow W=eQ_{H}

$$

The heat input rate is 1000 W. Assuming that the operational frequency of the cycle is $f_{cycle}=1\ \mathrm{Hz}$, thus leading to:

$$

Q_{H}=\frac{1000\ \mathrm{W}}{1\ \mathrm{Hz}}=1000 J

$$

So, the power output (work/cycle):

$$

P_{out}=W\cdot f_{cycle}=e_{C}\cdot Q_{H}\cdot f_{cycle}=0.5\cdot 1000\ \mathrm{J}\cdot 1\ \mathrm{Hz}=500\ \mathrm{W}

$$

2) The coefficient of performance (COP) of a heat pump is equal to 4. If the heat pump absorb 100 J of energy per cycle from a cold reservoir, how much work is required per cycle, and how much heat is expelled to the hot reservoir per cycle?


The coefficient of performance COP of a heat pump is defined as:

$$

\mathrm{COP}=\frac{|Q_{H}|}{|W|}=4,

$$

so the heat released to the hot reservoir is equal to:

$$

|Q_{H}|=\mathrm{COP}\cdot W=4W

$$

To dtermine the amount of heat expelled to the hot reservoir per cycle, let us recall the first law of thermodynamics in the case of cyclic processes:

$$

Q_{H}+Q_{C}=W

$$

, which results in, substituting the expression for $Q_{H}$:

$$

4W+Q_{C}=W\Rightarrow 3W=-100\ \mathrm{J}\Rightarrow W=-33.3\ \mathrm{J}

$$

, and then:

$$

Q_{H}=4W=-133.3\ \mathrm{J}

$$


3) (*UPMP* 20.9): The Otto-cycle engine in a Porsche 911 Carrera has a compression ratio of 10.2.

    (a) What is the ideal efficiency of the engine? Use $\gamma=1.40$. 
 
    (b) The engine in a Ferrari F430 has a slightly higher compression ratio of 11.3. How much increase in the ideal efficiency results from this increase in the compression ratio?

(a) The efficiency of the Otto-cycle can be expressed as:

$$

e_{\mathrm{Otto}}^{\mathrm{Porsche}}=1-\frac{1}{r^{\gamma-1}}=1-\frac{1}{10.2^{1.40-1}}=0.605

$$

(b) In the case of the Ferrari, the efficiency is, using the same formula, equal to $e_{Otto}^{Ferrari}=0.621$, thus resulting in an increase:

$$

\frac{e_{\mathrm{Otto}}^{\mathrm{Ferrari}}-e_{\mathrm{Otto}}^{\mathrm{Porsche}}}{e_{\mathrm{Otto}}^{\mathrm{Porsche}}}=0.026\Rightarrow 2.6 \%

$$

(b) The compression ratio of the engine is a quantity associated to the the fuel mixture capacity of being compressed inside the chamber before the ignition. It is often expressed as $r$ and depends on the quality of the used fuel. A key parameter to classify fuels and their degree of compressibility is the number of octanes, which indicates the the resistance of the air-fuel mixture to explode. An high number of octanes-fuel translates in a more compressible mixture.

(c) Diesel engines don’t use spark plugs because they ignite fuel using high compression instead of a spark. Differently from Otto-cycle, where the air-fuel mixture is first compressed and then ignited by means of a spark plug, in Diesel cycle fuel is injected inside the chamber after the compression stroke of air. This allows to reach higher compressions without risking auto-ignition. Once diesel fuel is injected into hot compressed air, it auto-ignites.
This is the reason why Diesel cycle reaches higher thermal efficiency than Otto-cycle.

6) **Question 6**(*UPMP* 20.34): A heat engine takes 0.350 mol of a diatomic ideal gas around the cycle shown in the $pV$-diagram of {numref}`fig:question`. Process $1\rightarrow 2$ is at constant volume, process $2\rightarrow 3$ is adiabatic, and process $3\rightarrow 1$ is at a constant pressure of 1.00 atm. The value of $\gamma$ for this gas is 1.40.

    (a)Find the pressure and volume at points 1,2 and 3.

    (b)Calculate $Q,W$ and $\Delta U$ for each of the three processes.

    (c)Find the net work done by the gas in the cycle.

    (d)Find the net heat flow into the engine in cycle.
    
    (e)What is the thermal efficiency of the engine? How does this compare to the efficiency of a Carnot-cycle engine operating between the same minimum and maximum temperatures $T_{1}$ and $T_{2}$?




(a) Considering state 1 of the cyclic process (and remembering that $1\ \mathrm{atm}=1.01\cdot10^{5}\ \mathrm{Pa}$):

$$

V_{1}=\frac{nRT_{1}}{p_{1}}=\frac{0.350\ \mathrm{mol}\cdot 8.314\ \frac{\mathrm{J}}{\mathrm{mol}\cdot\mathrm{K}}\cdot 300\ \mathrm{K}}{1.01\cdot 10^{5}\ \mathrm{Pa}}=8.64\cdot 10^{-3}\ \mathrm{m}^{3}

$$

Considering state 2, we observe $V_{2}=V_{1}$ as the process in isochoric, and:

$$

p_{2}=\frac{nRT_{2}}{V_{2}}=\frac{0.350\ \mathrm{mol}\cdot 8.314\ \frac{\mathrm{J}}{\mathrm{mol}\cdot\mathrm{K}}\cdot 600\ \mathrm{K}}{8.026\cdot 10^{-3}\ \mathrm{m}^{3}}=2.02\cdot 10^{5}\ \mathrm{Pa}

$$

To calculate volume in state 3, let us remember that for a diatomic gas $c_{V}=\frac{5}{2}R$, $c_{P}=\frac{7}{2}R$ and then $\gamma=\frac{7}{5}=1.4$. Moreover, it is important to observe that $p_{3}=p_{1}$: in this way it is possible to derive:

$$

p_{2}V_{2}^{\gamma}=p_{3}V_{3}^{\gamma}\Rightarrow V_{3}=V_{2}\cdot ^{\gamma}\sqrt{\frac{p_{2}}{p_{3}}},

$$

which then results in:

$$

V_{3}=8.026\cdot 10^{-3}\ \mathrm{m}^{3} \cdot ^{1.4}\sqrt{\frac{2.02\cdot 10^{5}\ \mathrm{Pa}}{1.01 \cdot 10^{5}\ \mathrm{Pa}}}=14.17\cdot 10^{-3}\ \mathrm{m}^{3}

$$


(b) Let us start from transformation $1\rightarrow 2$:

$$

W_{12}=0,\quad Q_{12}=\Delta U= nc_{V}\Delta T=0.350\ \mathrm{mol}\cdot \frac{5}{2}R\cdot \left(600\ \mathrm{K}-300 \mathrm{K}\right)=2182.43\ \mathrm{J}

$$

Consider now transformation $2\rightarrow 3$:

$$

Q_{23}=0

$$

$$

\begin{aligned}
W_{23}&=-\Delta U=-nc_{V}\Delta T=\\
&=0.350\ \mathrm{mol}\cdot \frac{5}{2}R\cdot \left(492\ \mathrm{K}-600 \mathrm{K}\right)=\\
&=785.67\ \mathrm{J}
\end{aligned}

$$

From transformation $3\rightarrow 1$ instead, it results:

$$

\begin{aligned}
W_{31}&=p\Delta V=\\
&=1.01\cdot 10^{5}\ \mathrm{Pa}\cdot \left(8.64\cdot 10^{-3}\ \mathrm{m}^{3}-14.17\cdot 10^{-3}\ \mathrm{m}^{3}\right)=\\
&=-558.53\ \mathrm{J}
\end{aligned}

$$

$$

\begin{aligned}
Q_{31}&=\Delta U+W=nc_{V}\Delta T+W=\\
&=0.350\ \mathrm{mol}\cdot \frac{5}{2}R\cdot \left(300\ \mathrm{K}-492 \mathrm{K}\right)-558.53\ \mathrm{J}=\\
&=\underbrace{-1396.75\ \mathrm{J}}_{\Delta U}-558.53\ \mathrm{J}=\\
&=-1955.28\ \mathrm{J}
\end{aligned}

$$

(c) The net work done by the gas in the cycle corresponds to:

$$

\begin{aligned}
W_{tot}&=W_{12}+W_{23}+W_{31}=\\
&=\left(0+785.67 J-558.53\right)\ \mathrm{J}=\\
&=227.14\ \mathrm{J}
\end{aligned}

$$

(d) The net heat flow into the engine cycle is equal to:

$$

\begin{aligned}
Q_{tot}&=Q_{12}+Q_{23}+Q_{31}=\\
&=\left(2182.43+0-1955.28\right)\ \mathrm{J}=\\
&=227.14\ \mathrm{J}
\end{aligned}

$$

, which is characteristic of cyclic processes.

(e) The thermal efficiency of the engine is:

$$

e=\frac{W_{tot}}{Q_{H}}=\frac{W_{tot}}{Q_{12}}=0.104

$$

The associated Carnot cycle operating between 300 K and 600 K has an efficiency:

$$

e_{C}=\frac{600\ \mathrm{K}-300\ \mathrm{K}}{600\ \mathrm{K}}=0.5

$$

, which confirms that the Carnot engine is the most efficient engine.

7) **Question 7**: Explain with the aid of diagrams the operation of the Carnot cycle, detailing each stage in the process.

The answer can be found in Module 6.2 of the Jupyter Notebook or chapter 20.6 of UPMP
