9. A physical system has a lifetime that follows an exponential distribution, such as the lifetime of a radioactive nucleus. The probability density is

   $$
   P(t) =
   \begin{cases}
   \lambda e^{-\lambda t}, & t \ge 0, \\
   0, & t < 0,
   \end{cases}
   $$

   where $\lambda$ is the decay rate constant.

   1. Show that this probability distribution is normalised.
   2. Calculate the mean lifetime, $\langle t \rangle$.
   3. Calculate the second moment, $\langle t^2 \rangle$.
   4. Show that the variance is $\sigma_t^2 = \frac{1}{\lambda^2}$, and comment on what this means physically.
