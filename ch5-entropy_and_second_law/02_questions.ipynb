   2. Using $\Delta S = k_B \ln \Omega_2 - k_B \ln \Omega_1 = k_B \ln \frac{\Omega_2}{\Omega_1}$, calculate the change in entropy of the gas.
   3. Compare this to the entropy change for a free expansion between the same states, where $\Delta S = \frac{Q}{T}$ for a reversible process. What do you notice?

7. A block of lead of heat capacity 1 kJ K$^{-1}$ is cooled from 200 K to 100 K in two ways. (TP 14.5)
   1. It is plunged into a large liquid bath at 100 K.
   2. The block is first cooled to 150 K in one liquid bath and then to 100 K in another bath.

   Calculate the entropy changes in the system comprising block plus baths in each case. Prove that in the limit of an infinite number of intermediate baths the total entropy change is zero.
