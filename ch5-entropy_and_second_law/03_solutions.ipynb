### Solutions

1. Two moles of an ideal gas occupy a volume $V$. The gas expands isothermally and reversibly to a volume $3V$.
    1. Is the velocity distribution changed by the isothermal expansion? Explain.

      <div class="answer-narrow">

      No. Since the expansion is isothermal, T remains constant, so the distribution of molecular speeds is unchanged.
      
			</div>

    2. Using $\Delta S = \int \frac{dQ_{\text{rev}}}{T} = nR\ln\frac{V_f}{V_i}$, calculate the change in entropy of the gas.

      <div class="answer-narrow">

      $$\Delta S = nR\ln\frac{V_f}{V_i} = (2)(8.314)\ln\frac{3V}{V} = 2R\ln 3 = 18.3 \text{ J K}^{-1}$$

      </div>

    3. Compare this to the entropy change for a free expansion between the same states, where $\Delta S = nR\ln\frac{V_f}{V_i}$. What do you notice?

      <div class="answer-narrow">

      The result is identical: $\Delta S = 18.3$ J K$^{-1}$. This is expected — entropy is a state function, so $\Delta S$ depends only on the initial and final states, not on whether the path is reversible or irreversible.

      </div>

5. A mug of tea has been left to cool from 90$^\circ C$ to 18$^\circ C$. If there is 0.2 kg of tea in the mug, and the tea has specific heat capacity 4200 J K$^{-1}$ kg$^{-1}$, show that the entropy of the tea has decreased by 185.7 J K$^{-1}$. Comment on the sign of this result.

    <div class="answer-narrow">
    
    $\Delta S = -\int_{291}^{363} C \frac{dT}{T} = -C \ln(363/291) = -185.7$ J K$^{-1}$. The sign is negative because the tea loses heat. However, the entropy of the surroundings increases by $C(363 - 291)/291 = +207.8$ J K$^{-1}$, which is larger, so the entropy of the Universe increases overall.

    </div>

6. In a free expansion of a perfect gas (also called Joule expansion), we know $U$ does not change, and no work is done. However, the entropy must increase because the process is irreversible. Are these statements compatible with the first law $dU = TdS - pdV$?

    <div class="answer-narrow">

    Yes: the first law always holds. In a free expansion $dU = 0$ and $dW = pdV = 0$, so $TdS = pdV$, meaning both sides are zero and the equation is satisfied. The entropy increase is real but is calculated via a reversible path between the same endpoints.

    </div>

7. A block of lead of heat capacity 1 kJ K$^{-1}$ is cooled from 200 K to 100 K in two ways.
   1. It is plunged into a large liquid bath at 100 K.

    <div class="answer-narrow">

    $\Delta S_\mathrm{total} = -C_V \ln 2 + C_V = C_V(1 - \ln 2) = 0.307$ J K$^{-1}$

    </div>

   2. The block is first cooled to 150 K in one liquid bath and then to 100 K in another bath.

    <div class="answer-narrow">

    $\Delta S_\mathrm{total} = 0.14$ J K$^{-1}$
    In the limit of an infinite number of intermediate baths, $\Delta S_\mathrm{total} \to 0$, since the process becomes reversible.

    </div>
