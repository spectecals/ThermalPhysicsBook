5. A mug of tea has been left to cool from 90$^\circ C$ to 18$^\circ C$. If there is 0.2 kg of tea in the mug, and the tea has specific heat capacity 4200 J K$^{-1}$ kg$^{-1}$, show that the entropy of the tea has decreased by 185.7 J K$^{-1}$. Comment on the sign of this result.

    <div class="answer-narrow">
    
    $\Delta S = -\int_{291}^{363} C \frac{dT}{T} = -C \ln(363/291) = -185.7$ J K$^{-1}$. The sign is negative because the tea loses heat. However, the entropy of the surroundings increases by $C(363 - 291)/291 = +207.8$ J K$^{-1}$, which is larger, so the entropy of the Universe increases overall.

    </div>

6. In a free expansion of a perfect gas (also called Joule expansion), we know $U$ does not change, and no work is done. However, the entropy must increase because the process is irreversible. Are these statements compatible with the first law $dU = TdS - pdV$?

    <div class="answer-narrow">

    Yes: the first law always holds. In a free expansion $dU = 0$ and $dW = pdV = 0$, so $TdS = pdV$, meaning both sides are zero and the equation is satisfied. The entropy increase is real but is calculated via a reversible path between the same endpoints.

    </div>

7. A 10 $\Omega$ resistor is held at a temperature of 300 K. A current of 5 A is passed through the resistor for 2 minutes. Ignoring changes in the source of the current, what is the change of entropy in:
   1. the resistor?

    <div class="answer-narrow">

    $\Delta S = 0$, because $T$ is constant (the resistor is held at 300 K, so heat flows out as fast as it is generated).

    </div>

   2. the Universe?

    <div class="answer-narrow">

    $I^2 R t = 3 \times 10^4$ J flows into the environment. Hence $\Delta S = 3 \times 10^4 / 300 = 100$ J K$^{-1}$.

    </div>

9. A block of lead of heat capacity 1 kJ K$^{-1}$ is cooled from 200 K to 100 K in two ways.
   1. It is plunged into a large liquid bath at 100 K.

    <div class="answer-narrow">

    $\Delta S_\mathrm{total} = -C_V \ln 2 + C_V = C_V(1 - \ln 2) = 0.307$ J K$^{-1}$

    </div>

   2. The block is first cooled to 150 K in one liquid bath and then to 100 K in another bath.

    <div class="answer-narrow">

    $\Delta S_\mathrm{total} = 0.14$ J K$^{-1}$
    In the limit of an infinite number of intermediate baths, $\Delta S_\mathrm{total} \to 0$, since the process becomes reversible.

    </div>

10. Calculate the changes in entropy of the Universe as a result of the following processes:

    1. A capacitor of capacitance 1 $\mu$F is connected to a battery of emf 100 V at 0$^\circ C$.

    <div class="answer-narrow">

    $\Delta S_\mathrm{Universe} = \frac{1}{2}CV^2/273 = 18.3$ $\mu$J K$^{-1}$

    </div>

    2. The same capacitor, after being charged to 100 V, is discharged through a resistor at 0$^\circ C$.

    <div class="answer-narrow">

    $\Delta S_\mathrm{Universe} = \frac{1}{2}CV^2/273 = 18.3$ $\mu$J K$^{-1}$

    </div>

    3. One mole of gas at 0$^\circ C$ is expanded reversibly and isothermally to twice its initial volume.

    <div class="answer-narrow">

    $\Delta S_\mathrm{gas} = R\ln 2$, $\Delta S_\mathrm{surr} = -R\ln 2$, so $\Delta S_\mathrm{Universe} = 0$

    </div>
    
    4. One mole of gas at 0$^\circ C$ is expanded reversibly and adiabatically to twice its initial volume.

    <div class="answer-narrow">

    $dQ = 0$ so $\Delta S_\mathrm{gas} = \Delta S_\mathrm{surr} = 0$, hence $\Delta S_\mathrm{Universe} = 0$

    </div>
    
    5. The same expansion as in (iv) is carried out by opening a valve to an evacuated container of equal volume.
    
    <div class="answer-narrow">

    Joule expansion: $\Delta S_\mathrm{gas} = R\ln 2$, $\Delta S_\mathrm{surr} = 0$, so $\Delta S_\mathrm{Universe} = R\ln 2 = 5.76$ J K$^{-1}$

    </div>

11. Consider $n$ moles of a gas, initially confined within a volume $V$ and held at temperature $T$. The gas is expanded to a total volume $\alpha V$.
    1. Reversible isothermal expansion:
    
    <div class="answer-narrow">

    $\Delta S = nR \ln \alpha$

    2. Free expansion into vacuum:
    
    <div class="answer-narrow">

    $\Delta S = nR \ln \alpha$ (same value; $S$ is a function of state)
    For the van der Waals gas in case (i): $\Delta S = nR \ln\left(\frac{\alpha V - nb}{V - nb}\right)$. For case (ii) the temperature falls by $\Delta T = -\frac{an^2}{C_V V}\left(\frac
    {\alpha - 1}{\alpha}\right)$.

12. The probability of a system being in the $i$th microstate is $P_i = e^{-\beta E_i}/Z$. Show that the entropy is given by $S/k_B = \ln Z + \beta U$.

    <div class="answer-narrow">

    $S/k_B = -\sum_i P_i \ln P_i = -\sum_i P_i(-\beta E_i - \ln Z) = \ln Z + \beta U$

    </div>

13. Use the Gibbs expression for entropy $S = -k_B \sum_i P_i \ln P_i$ to derive the formula for the entropy of mixing.

    <div class="answer-narrow">

    With $N$ molecules and probability $x$ that a molecule is type 1 and probability $(1-x)$ that it is type 2: $S / Nk_B = -x\ln x - (1-x)\ln(1-x)$

    </div>
