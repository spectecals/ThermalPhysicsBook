### Solutions

1. Two moles of an ideal gas occupy a volume $V$. The gas expands isothermally and reversibly to a volume $3V$.


   1. Is the velocity distribution changed by the isothermal expansion? Explain.

    <div class="answer-narrow">

    No. The Maxwell-Boltzmann speed distribution depends only on temperature, which is unchanged in an isothermal process.

    </div>

   2. Calculate the change in entropy of the gas.

    <div class="answer-narrow">

    $\Delta S = nR\ln(V_f/V_i) = 2R\ln 3 = 18.3$ J K$^{-1}$

    </div>

   3. Compare this to the change in entropy for a free expansion between the same states.

    <div class="answer-narrow">

    18.3 J K$^{-1}$ — identical, since entropy is a state function and does not depend on the path taken between states.

    </div>

2. 0.35 kg of ice at 0.0$^\circ$C is heated until it is all melted.


   1. What is the change in entropy of this body?

    <div class="answer-narrow">

    $\Delta S = mL_f/T = (0.35)(334\times10^3)/273 = +429$ J K$^{-1}$

    </div>

   2. The source of heat is a very massive body at a temperature of 25.0$^\circ$C. What is the change in entropy of this body?

    <div class="answer-narrow">

    $\Delta S = -Q/T = -(0.35)(334\times10^3)/298 = -393$ J K$^{-1}$

    </div>

   3. What is the total change in entropy of the water and the heat source?

    <div class="answer-narrow">

    $\Delta S_\mathrm{total} = 429 - 393 = +36$ J K$^{-1}$

    </div>

3. Three moles of an ideal gas undergo a reversible isothermal compression at 20.0$^\circ$C. During this compression, 1850 J of work is done on the gas. What is the change of entropy of the gas?

  <div class="answer-narrow">

  For a reversible isothermal process $\Delta U = 0$, so $Q = -W = -1850$ J. Hence

  $$\Delta S = Q/T = -1850/293 = -6.31 \text{ J K}^{-1}$$

  </div>

4. A 10.0 L gas tank containing 3 moles of ideal He gas at 20.0$^\circ$C is placed inside a completely evacuated, insulated bell jar of volume 35.0 L. A small hole in the tank allows the He to leak out into the jar until the gas reaches a final equilibrium state with no more leakage.


   1. What is the change in entropy of this system due to the leaking of the gas?

    <div class="answer-narrow">

    Free expansion into vacuum: $\Delta S = nR\ln(V_f/V_i) = 3R\ln(35/10) = +33.3$ J K$^{-1}$

    </div>

   2. Is the process reversible or irreversible? How do you know?

    <div class="answer-narrow">

    Irreversible: the entropy of the system increases ($\Delta S > 0$) while the surroundings are insulated ($\Delta S_\mathrm{surr} = 0$), so $\Delta S_\mathrm{Universe} > 0$.

    </div>

5. A mug of tea has been left to cool from 90$^\circ$C to 18$^\circ$C. If there is 0.2 kg of tea in the mug, and the tea has specific heat capacity 4200 J K$^{-1}$ kg$^{-1}$, show that the entropy of the tea has decreased by 185.7 J K$^{-1}$. Comment on the sign of this result.

  <div class="answer-narrow">

  $$\Delta S = \int_{363}^{291} \frac{mc\,dT}{T} = mc\ln\frac{291}{363} = (0.2)(4200)\ln\frac{291}{363} = -185.7 \text{ J K}^{-1}$$

  The sign is negative because the tea loses heat. The entropy of the surroundings increases by $mc(363-291)/291 = +207.8$ J K$^{-1}$, which is larger, so the entropy of the Universe increases — consistent with the second law.

  </div>

  <br><br>

6. In a free expansion of a perfect gas (also called Joule expansion), we know $U$ does not change, and no work is done. However, the entropy must increase because the process is irreversible. Are these statements compatible with the first law $dU = TdS - pdV$?

  <div class="answer-narrow">

  Yes. The first law always holds. In a free expansion $dU = 0$ and $\bar{d}W = p\,dV = 0$, so $TdS = p\,dV$, meaning both sides are zero — the equation is satisfied. The entropy increase is real but is computed via a reversible path connecting the same initial and final states.

  </div>

7. A 10 $\Omega$ resistor is held at a temperature of 300 K. A current of 5 A is passed through the resistor for 2 minutes. Ignoring changes in the source of the current, what is the change of entropy in:


   1. the resistor?

    <div class="answer-narrow">

    $\Delta S = 0$: the resistor is held at constant temperature, so heat flows out as fast as it is generated and no entropy accumulates in it.

    </div>

   2. the Universe?

    <div class="answer-narrow">

    $Q = I^2Rt = (25)(10)(120) = 3\times10^4$ J flows into the environment. Hence $\Delta S = Q/T = 3\times10^4/300 = 100$ J K$^{-1}$.

    </div>

    <br><br>
    </RAW HTML BREAK>

8. Calculate the change of entropy:


   1. of a bath containing water, initially at 20$^\circ$C, when it is placed in thermal contact with a very large heat reservoir at 80$^\circ$C,

    <div class="answer-narrow">

    $\Delta S_\mathrm{bath} = C\ln(T_f/T_i) = 10^4\ln(353/293) = +1.9\times10^3$ J K$^{-1}$

    </div>

   2. of the reservoir when process (i) occurs,

    <div class="answer-narrow">

    $\Delta S_\mathrm{res} = -C(T_f - T_i)/T_\mathrm{res} = -10^4(353-293)/353 = -1.7\times10^3$ J K$^{-1}$

    </div>

   3. of the bath and of the reservoir if the bath is brought to 80$^\circ$C through the operation of a Carnot engine between them.

    <div class="answer-narrow">

    Zero: a Carnot engine is reversible, so $\Delta S_\mathrm{Universe} = 0$.

    </div>

9. A block of lead of heat capacity 1 kJ K$^{-1}$ is cooled from 200 K to 100 K in two ways:


   1. it is plunged into a large liquid bath at 100 K,

    <div class="answer-narrow">

    $\Delta S_\mathrm{total} = -C_V\ln 2 + C_V = C_V(1 - \ln 2) = 0.307$ J K$^{-1}$

    </div>

   2. the block is first cooled to 150 K in one liquid bath and then to 100 K in another.

    <div class="answer-narrow">

    $\Delta S_\mathrm{total} = -C_V\ln 2 + C_V(\frac{1}{3} + \frac{1}{2}) = 0.14$ J K$^{-1}$

    In the limit of infinitely many intermediate baths the process becomes reversible and $\Delta S_\mathrm{total} \to 0$.

    </div>

    <br><br>
    </RAW HTML BREAK>

10. Calculate the changes in entropy of the Universe as a result of the following processes:


   1. A capacitor of capacitance 1 $\mu$F is connected to a battery of emf 100 V at 0$^\circ$C.

    <div class="answer-narrow">

    Half the energy $\frac{1}{2}CV^2$ is dissipated as heat in the battery. $\Delta S_\mathrm{Universe} = \frac{1}{2}CV^2/T = \frac{1}{2}(10^{-6})(10^4)/273 = 18.3\ \mu$J K$^{-1}$

    </div>

   2. The same capacitor, after being charged to 100 V, is discharged through a resistor at 0$^\circ$C.

    <div class="answer-narrow">

    The stored energy $\frac{1}{2}CV^2$ is dissipated in the resistor. $\Delta S_\mathrm{Universe} = \frac{1}{2}CV^2/T = 18.3\ \mu$J K$^{-1}$

    </div>

   3. One mole of gas at 0$^\circ$C is expanded reversibly and isothermally to twice its initial volume.

    <div class="answer-narrow">

    Reversible: $\Delta S_\mathrm{gas} = R\ln 2$, $\Delta S_\mathrm{surr} = -R\ln 2$, so $\Delta S_\mathrm{Universe} = 0$.

    </div>

   4. One mole of gas at 0$^\circ$C is expanded reversibly and adiabatically to twice its initial volume.

    <div class="answer-narrow">

    $\bar{d}Q = 0$ so $\Delta S_\mathrm{gas} = 0$ and $\Delta S_\mathrm{Universe} = 0$.

    </div>

   5. The same expansion as in (iv) is carried out by opening a valve to an evacuated container of equal volume.

    <div class="answer-narrow">

    Joule (free) expansion: $\Delta S_\mathrm{gas} = R\ln 2$, $\Delta S_\mathrm{surr} = 0$, so $\Delta S_\mathrm{Universe} = R\ln 2 = 5.76$ J K$^{-1}$.

    </div>

11. Consider $n$ moles of a gas, initially confined within a volume $V$ and held at temperature $T$. The gas is expanded to a total volume $\alpha V$, where $\alpha$ is a constant, by:


   1. a reversible isothermal expansion, and

    <div class="answer-narrow">

    $\Delta S = nR\ln\alpha$

    For a van der Waals gas: $\Delta S = nR\ln\!\left(\dfrac{\alpha V - nb}{V - nb}\right)$

    </div>

   2. removing a partition and allowing a free expansion into the vacuum.

    <div class="answer-narrow">

    $\Delta S = nR\ln\alpha$ — identical to (i), since $S$ is a state function and is independent of path.

    For a van der Waals gas the temperature falls by $\Delta T = -\dfrac{an^2}{C_V V}\left(\dfrac{\alpha-1}{\alpha}\right)$, which is proportional to $(\alpha-1)/\alpha$.

    </div>

12. The probability of a system being in the $i$th microstate is $P_i = e^{-\beta E_i}/Z$. Show that the entropy is given by $S/k_B = \ln Z + \beta U$.

  <div class="answer-narrow">

  $$S/k_B = -\sum_i P_i\ln P_i = -\sum_i P_i(-\beta E_i - \ln Z) = \ln Z\sum_i P_i + \beta\sum_i P_i E_i = \ln Z + \beta U$$

  </div>

13. Use the Gibbs expression for entropy $S = -k_B \sum_i P_i \ln P_i$ to derive the formula for the entropy of mixing.

  <div class="answer-narrow">

  With $N$ molecules and probability $x$ that a molecule is type 1 and probability $(1-x)$ that it is type 2:

  $$\frac{S}{Nk_B} = -x\ln x - (1-x)\ln(1-x)$$

  </div>
